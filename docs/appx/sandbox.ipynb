Конструкция вида `{span .text-red-500}` – это официальный синтаксис встроенных инлайн-атрибутов ([Inline Attributes](https://mystmd.org/guide/inline-options)) в **современной спецификации MyST Markdown (MyST CLI)**.

::::{note} Решение траблы через `custom.css`
:class: simple dropdown
:open: false

Создадим и подключим файл `custom.css` **с нуля, пошагово и строго по [официальной спецификации MyST Markdown](https://mystmd.org/guide/website-style)**.

---
**Шаг 1. Создание файла `custom.css`**

1. Открыть корень проекта (`Learning-SQL`) в VS Code или в Проводнике.
2. Создать новый текстовый файл и назвать **`custom.css`** (он должен лежать в той же папке, где находится файл `myst.yml`).
3. Вставить в него следующий код со всеми 17-ю цветами из таблицы (цвета взяты из оригинальной палитры Tailwind):

```css
/* ====================================================================
   Пользовательские цвета для роли {span}
   ==================================================================== */
.text-red-500     { color: #ef4444 !important; }
.text-red-600     { color: #dc2626 !important; }
.text-red-700     { color: #b91c1c !important; }

.text-blue-500    { color: #3b82f6 !important; }
.text-blue-600    { color: #2563eb !important; }

.text-green-500   { color: #22c55e !important; }

.text-orange-600  { color: #ea580c !important; }
.text-amber-600   { color: #d97706 !important; }
.text-yellow-600  { color: #ca8a04 !important; }

.text-indigo-400  { color: #818cf8 !important; }
.text-violet-200  { color: #ddd6fe !important; }
.text-purple-600  { color: #9333ea !important; }

.text-slate-500   { color: #64748b !important; }
.text-gray-500    { color: #6b7280 !important; }
.text-zinc-600    { color: #52525b !important; }
.text-neutral-700 { color: #404040 !important; }

```
4. Сохранить файл (`Ctrl + S`).

---
**Шаг 2. Подключение файла в `myst.yml` (Критически важный нюанс)**

По официальной спецификации MyST параметр для подключения файла стилей называется **`style:`** (а не `css:`) [Styles and CSS Classes](https://mystmd.org/guide/website-style), [Website Themes & Templates](https://mystmd.org/guide/website-templates).

Открыть файл **`myst.yml`** в корне проекта:
1. Найти в нем блок `site:`.
2. Внутрь блока добавить секцию `options:` и указать путь к файлу: `style: custom.css` [Многоразовый шаблон для создания веб-сайтов с использованием MyST Markdown](https://gishub.org/blog/myst-website-template/).

Пример, как должен выглядеть этот блок в `myst.yml`:

```yaml
site:
  template: book-theme
  options:
    style: custom.css
```
*(Если блок `options:` в `site:` уже есть, просто добавить строчку `style: custom.css` с правильным отступом в два пробела)*.

---
**Шаг 3. Полная очистка и запуск сборки**

Поскольку сборщик кэширует стили темы, изменения в файлах стилей не всегда подхватываются при обычном обновлении страницы.

В терминале (Git Bash или Anaconda Prompt в папке проекта) выполнить:

```bash
# 1. Стираем старый кэш сборки
jupyter book clean --all

# 2. Запускаем локальный веб-сервер
jupyter book start
```

---
**Шаг 4. Сброс кэша в браузере**

Браузеры (Chrome, Edge, Firefox) очень агрессивно сохраняют `.css` файлы в память. 

Когда локальный сайт откроется:
1. Перейти в раздел `Sandbox` на страницу с таблицей цветов.
2. Нажать комбинацию клавиш **`Ctrl + F5`** (жесткая перезагрузка страницы со сбросом кэша стилей).

Все 17 строк таблицы должны окраситься в соответствующие цвета, а синтаксис `{span .text-...}` останется стандартным и читаемым.
