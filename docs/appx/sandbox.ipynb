## div block

:::{topic} division block – секция (разделение, отделение)
Див-блок (или блок-контейнер) – служит _коробкой_ для группировки текста, картинок или других блоков.
:::

Шкала размеров текста в блоках `:::div` формируется через `:class:`

```yaml
:::{div}
:class: text-xs

`text-xs` – супер-мелкий (e**X**tra **S**mall)
:::
```

:::{div}
:class: text-xs
`text-xs` – супер-мелкий (e**X**tra **S**mall)
:::

:::{div}
:class: text-sm
`text-sm` – уменьшенный (**Sm**all)
:::

:::{div}
:class: text-base
`text-base` – обычный стандартный шрифт
:::

:::{div}
:class: text-lg
`text-lg` – слегка увеличенный (**L**ar**g**e)
:::

:::{div}
:class: text-xl
`text-xl` – крупный (e**X**tra **L**arge)
:::

:::{div}
:class: lead
`lead` – используется для лид-абзацев (1.25rem)
:::
