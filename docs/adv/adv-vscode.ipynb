### Размер шрифтов

По умолчанию в VS Code размер шрифта ячейки Markdown составляет 120% от размера шрифта ячейки Code. Из-за этого пояснительный текст выглядит непропорционально крупным на фоне блоков с кодом[^4].

[^4]: Непонятно чем думали дизайнеры VS Code когда решили, что текст в ячейках Markdown обязан *кричать* на фоне скромного блока с кодом. Эти дефолтные настройки бесят абсолютно всех авторов блокнотов.

Чтобы выровнять масштаб и сделать чтение блокнота визуально гармоничным: 

:::{card}
1. Открыть настройки, нажав {kbd}`Ctrl` + {kbd}`,` (запятая).
2. В строке поиска ввести: `font size`.
3. В первом пункте **Editor: Font Size** посмотреть текущее число (по умолчанию в Windows это обычно **14** px). 
4. В строке поиска ввести: `notebook markup font size`.
5. В поле **Notebook › Markup: Font Size** указать точно такое же число (**14**) вместо нуля.
:::

## Установка тегов видимости

Наконец добрались до главного – ради чего держим VS Code вторым редактором проекта.

Для управления видимостью ячеек на сайте используются метаданные блокнота. Чтобы назначить в метаданные ячейки тег видимости, в VS Code достаточно установить три расширения:

- [x] [Python](https://marketplace.visualstudio.com/items?itemName=ms-python.python) – базовый набор для работы с языком Python
- [x] [Jupyter](https://marketplace.visualstudio.com/items?itemName=ms-toolsai.jupyter) – набор поддержки файлов блокнотов (`.ipynb`)
    - [x] [Jupyter Cell Tags](https://marketplace.visualstudio.com/items?itemName=ms-toolsai.vscode-jupyter-cell-tags) – **инструмент** для добавления тегов видимости

В проекте используются три стандартных тега (для ячеек кода):

| Тег                | Что отображается на сайте                                                       |
| :----------------- | :------------------------------------------------------------------------------ |
| **`remove-input`** | Код **скрыт**, вывод ячейки (таблица / результат) **виден**                     |
| **`remove-cell`**  | Вся ячейка **полностью невидима**  (скрыты и код, и вывод)                      |
| **`hide-cell`**    | Вся ячейка (код и вывод) **свернута** в интерактивный спойлер **Notebook Cell** |

:::{note} Примечание

Теги видимости работают только при сборке веб-страниц. Они **не удаляют** код из файлов `.ipynb`. Локально блокнот остается на 100% прежним.
:::

Чтобы установить тег (например, `remove-input`):

:::{card}
1. Открыть в VS Code блокнот `.ipynb`.
2. Выделить нужную ячейку кода.
3. На панели инструментов в правом верхнем углу ячейки нажать **три точки ⋯ (More Actions...)**.
4. В выпадающем меню выбрать пункт **Add Cell Tag**.
5. Вверху экрана появится поле ввода. Ввести туда: `remove-input` и нажать {kbd}`Enter`.
:::

В левом нижнем углу ячейки появится добавленный тег **`✕ remove-input`**  и кнопка **`✛ Tag`** для быстрого добавления последующих меток.

Безусловно, прописать теги можно и в JupyterLab. \

1. Открыть в JupyterLab блокнот `.ipynb`.

:::{code} json
{
    "trusted": false,
    "tags": [
        "remove-input"
    ]
}
:::
6. Нажать галочку (**Commit changes to data**) для сохранения изменений.

(using-cell-tags)=
## Использование тегов на практике

Все началось с неожиданного откровения: JupySQL {ref}`исправно подсвечивал синтаксис <highlight-magic-sql>` SQL-кода в локальном блокноте Jupyter, но на собранном сайте подсветка пропала:

:::{code} python
%%sql
SELECT
    customer_id,
    COUNT(*) AS tot_rentals,
    SUM(amount) AS tot_payments
FROM payment
GROUP BY customer_id;
:::

*То есть при сборке сайта генератор Jupyter Book не распознает магическую команду `%%sql` как чистый SQL и выводит код без подсветки синтаксиса* [^5].

[^5]: В принципе, в книге Алана Болье (оригинальном печатном издании) SQL-код не имеет подсветки синтаксиса. Но раз уж в локальных блокнотах подсветка есть, то мне захотелось понять как реализовать ее и на сайте.

---

(remove-input)=
### remove-input: подсветка синтаксиса

Чтобы на сайте получить цветной SQL-код с сохранением реального результата запроса, есть трюк со связкой двух ячеек:
- Прямо перед исполняемой ячейкой `%%sql` необходимо добавить текстовую ячейку Markdown c блоком кода ` ```sql `.
- Саму исполняемую ячейку {ref}`скрыть с помощью тега <set-cell-tags>` `remove-input`.

В результате на веб-странице останется *цветной* блок SQL-кода из ячейки Markdown, а сразу под ним – таблица результатов, возвращенная скрытой ячейкой `%%sql`:

```sql
SELECT
    customer_id,
    COUNT(*) AS tot_rentals,
    SUM(amount) AS tot_payments
FROM payment
GROUP BY customer_id;
```

`````python
%%sql
SELECT
    customer_id,
    COUNT(*) AS tot_rentals,
    SUM(amount) AS tot_payments
FROM payment
GROUP BY customer_id;
`````

Этот прием продемонстрировал во 2-й главе и точечно в других главах (например, в подразделе {ref}`Обобщенные табличные выражения <cte>` Главы 9).


:::{card} Резюме

- Если хотите с помощью Jupyter Book создать интерактивную веб-книгу с аккуратным подсвеченным синтаксисом SQL-запросов – этот прием для вас.

- Если ведете личный конспект – извращаться с дублированием кода и установкой тегов незачем: локальной подсветки в ячейках `%%sql` более чем достаточно.
:::

**Любопытный факт**:

- В редакторе JupyterLab: плагин разметки [jupyterlab_myst](https://jupyter-book.github.io/jupyterlab-myst/) не понимает как подсветить синтаксис SQL в ячейке Markdown и оставляет блоки ` ```sql ` монохромными, хотя при сборке сайта подсветка генерируется штатно.

- В VS Code такого визуального бага нет: подсветка синтаксиса одинаково корректно работает как в кодовой ячейке `%%sql`, так и в ячейке Markdown.

Поведение подсветки SQL в зависимости от типа ячейки и среды:

| Тип ячейки                     | Веб-сайт | JupyterLab | VS Code |
| :----------------------------- | :------- | :--------- | :------ |
| **Исполняемая `%%sql`**        | Нет      | Есть       | Есть    |
| **Блок ` ```sql ` в Markdown** | Есть     | Нет        | Есть    |


(remove-cell)=
### remove-cell: книжный стандарт

Тег **`remove-cell`** полностью скрывает ячейку со страницы сайта: не отображается ни сам код, ни результат его выполнения.

В обычной книге сразу за заголовком главы начинается изложение темы. В блокнотах Jupyter первой строкой идет технический блок – [ячейка подключения к базе данных](adv-sqlalchemy-jupysql.ipynb#connection-init). Без нее блокнот не сможет выполнить запросы.

Если цель – создать интерактивную книгу, то читателю не нужно видеть этот служебный код (импорт библиотек, создание Engine SQLAlchemy, пароли и статус соединения) в начале каждой главы[^6].

[^6]: Повторение одинакового блока подключения в каждой главе перегружает первый экран страницы визуальным шумом вместо погружения в тему главы.

Тег `remove-cell` как раз решает эту задачу: в блокноте ячейка остается на месте и штатно подключается к базе, а со страницы сайта она полностью скрыта.

:::{card} Наглядный пример
- Именно так на сайте выглядят главы с 9 по 11. Если открыть, например, [Главу 11](/ch11) – изложение темы начинается сразу под заголовком.

- Более того, **на этой самой странице** блока инициализации тоже нет. Однако демонстрационный SQL-запрос в подразделе выше выполнился и вывел таблицу результатов. Эту задачу решила первая ячейка, скрытая с помощью `remove-cell`.
:::

(hide-cell)=
### hide-cell: прячем под спойлер

Если `remove-cell` скрывает ячейку полностью, то `hide-cell` сворачивает ее в интерактивный спойлер **Notebook Cell**.

Читатель при необходимости нажимает на плашку спойлера и изучает код или вывод, а страница в свернутом виде остается компактной и аккуратной:

`````python
import pandas as pd
import pymysql
import sqlalchemy as sa
import sql

pd.set_option('display.max_rows', 20)

%load_ext sql
%config SqlMagic.displaycon = False
%config SqlMagic.autopandas = True

connection_url = sa.engine.URL.create(
    drivername='mysql+pymysql',
    host='localhost',
    port=3306,
    database='sakila',
    username='root',
    password='********',
)
engine = sa.create_engine(connection_url)

%sql engine

print(f"Pandas ver. {pd.__version__}: порог усечения строк уменьшен до 20")
print(f"SQLAlchemy ver. {sa.__version__}: подключение создано")
print(f"PyMySQL ver. {pymysql.__version__}: DBAPI-драйвер подключения")
print(f"JupySQL ver. {sql.__version__}: подключен через SQLAlchemy Engine")
`````

Этот компромиссный вариант используется в проекте для двух сценариев:

- **Первый** – оставить ячейку подключения на странице (пример выше реализован в главах 2–8).

::::{note} Почему нельзя указать `#| label:` прямо в коде?
:::{div}
:class: text-sm
В Python-ячейках MyST поддерживает метаданные через комментарий `#| label: sol-09-01` в первой строке. Но в наших блокнотах первая строка и первая колонка заняты магической командой `%%sql` – любой комментарий перед ней нарушит синтаксис IPython, и запрос перестанет выполняться. Поэтому для SQL-ячеек метка задается только через JSON-метаданные.
