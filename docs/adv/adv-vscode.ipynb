В качестве интерактивной среды разработки (IDE) в проекте используется JupyterLab. Выбор продиктован сугубо прикладными задачами интерактивных блокнотов (Notebooks):
- в них удобно писать теорию и тут же запускать SQL-запросы;
- файлы блокнотов `.ipynb` служат исходниками для сборки страниц сайта Jupyter Book;
- специфическая разметка MyST Markdown полноценно визуализируется (предпросматривается прямо в рабочем файле) только в среде Jupyter.

Однако некоторые сервисные операции оказалось удобнее выполнять в редакторе **Visual Studio Code** (далее VS Code):
- редактировать конфигурационные файлы проекта `myst.yml` и `toc.yml`;
- но **главное** – быстро проставлять теги MyST в метаданные ячеек через визуальный интерфейс, не открывая вручную *сырой* JSON-код блокнота.

Как бы ни хотелось использовать VS Code единственным редактором проекта, у него есть существенный недостаток: в отличие от JupyterLab он не умеет визуализировать сложную разметку MyST Markdown.

### Размер шрифтов

По умолчанию в VS Code размер шрифта ячейки Markdown составляет 120% от размера шрифта ячейки Code. Из-за этого пояснительный текст выглядит непропорционально крупным на фоне блоков с кодом.

Чтобы выровнять масштаб и сделать чтение блокнота визуально гармоничным: 

1. Открыть настройки, нажав `Ctrl + ,` (запятая).
2. В строке поиска ввести: `font size`.
3. В первом пункте **Editor: Font Size** посмотреть текущее число (по умолчанию в Windows это обычно **14** px). 
4. В строке поиска ввести: `notebook markup font size`.
5. В поле **Notebook › Markup: Font Size** указать точно такое же число (**14**) вместо нуля.

### Редактирование

В ходе работы вполне может случиться, что находясь в VS Code вы заметите опечатку и захотите внести правки сразу, не запуская JupyterLab.

Это нормальная ситуация. Главное – убедиться, что ячейка будет выполняться в окружении проекта. Для этого нужно всегда выбирать ядро с именем conda-окружения проекта:

1. Открыть любой файл `.ipynb`.
2. В правом верхнем углу окна нажать кнопку **Select Kernel**.
3. В появившемся списке (вверху, под полем поиска) выбрать окружение \
    `ds-book (Python 3.12.12) ~\anaconda3\envs\ds-book\python.exe`.
4. Название кнопки **Select Kernel** сменится на **ds-book (Python 3.12.12)**.
